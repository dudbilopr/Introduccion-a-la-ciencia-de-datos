## 1. Recolección: traer los datos
Los datos son reales: 731 días de alquileres de bicicletas en Washington D.C. (UCI, CC BY 4.0).

```python
df = pd.read_csv(BASE + "bicicletas_dia.csv", parse_dates=["dteday"])
print("Filas y columnas:", df.shape)
print("Celdas vacías:", df.isna().sum().sum())
df.head()
```

**Pregunta:** ¿qué representa una fila? ¿Y una columna?

## 2. Procesamiento: dejar los datos listos
La temperatura viene dividida por 41. La devolvemos a grados Celsius.

```python
df["temp_c"] = df["temp"] * 41
print("Temperatura media:", round(df["temp_c"].mean(), 1), "°C")
```

## 3. Análisis: describir
¿Cuántas bicicletas se alquilan en un día normal? ¿Cuál fue el día con menos alquileres?

```python
print("Promedio por día:", round(df["cnt"].mean()))
dia_min = df.loc[df["cnt"].idxmin()]
print("Día con menos alquileres:", dia_min["dteday"].date(), "con", dia_min["cnt"], "alquileres")
print("Clima ese día (1 despejado, 2 nublado, 3 lluvia o nieve ligera):", dia_min["weathersit"])
print("Temperatura ese día:", round(dia_min["temp_c"], 1), "°C")
```

## 4. Visualización: mirar el contexto
Graficamos octubre y noviembre de 2012.

```python
tramo = df[(df["dteday"] >= "2012-10-01") & (df["dteday"] <= "2012-11-30")]
print("Promedio de octubre de 2012:", round(tramo[tramo["dteday"].dt.month == 10]["cnt"].mean()))
plt.figure(figsize=(9, 3.5))
plt.bar(tramo["dteday"], tramo["cnt"], color=["#e85c0b" if v == tramo["cnt"].min() else "#5d8c96" for v in tramo["cnt"]])
plt.title("El 29 de octubre de 2012 casi nadie alquiló bicicleta")
plt.ylabel("Alquileres por día")
plt.show()
```

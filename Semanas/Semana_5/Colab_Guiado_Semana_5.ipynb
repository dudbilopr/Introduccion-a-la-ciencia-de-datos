```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
BASE = "https://raw.githubusercontent.com/dudbilopr/Introduccion-a-la-ciencia-de-datos/main/Semanas/Datos/"
```

Datos: 3.150 clientes de una empresa de telecomunicaciones (Iranian Churn, UCI, CC BY 4.0). Variables de los meses 1 a 9; `Churn` indica si el cliente se fue en el mes 12.

## 1. El expediente

```python
df = pd.read_csv(BASE + "churn_telecom.csv")
df.columns = [" ".join(c.split()) for c in df.columns]   # algunos nombres traen dos espacios
print(df.shape, " duplicadas:", df.duplicated().sum(), " se fue:", round(df["Churn"].mean() * 100, 1), "%")
print(df.columns.tolist())
```

## 2. Tres testigos del centro

```python
v = df["Customer Value"]
print("Media:", round(v.mean(), 2), " Mediana:", round(v.median(), 2), " Moda (intervalo de 50):", (v // 50 * 50).mode()[0])
sms = df["Frequency of SMS"]
print("SMS: media", round(sms.mean(), 2), " mediana", sms.median(), " sin SMS:", round((sms == 0).mean() * 100, 1), "%")
orden = np.sort(sms.to_numpy())[::-1]
print("El 10 % que más escribe envía el", round(orden[: len(orden) // 10].sum() / orden.sum() * 100, 1), "% de los SMS")
```

## 3. Interrogatorio: un cliente gigante

```python
extremo = np.append(v.to_numpy(), v.max() * 100)
print("Media:", round(v.mean(), 1), "->", round(extremo.mean(), 1))
print("Mediana:", round(v.median(), 1), "->", round(np.median(extremo), 1))
print("Desviación estándar:", round(v.std(), 1), "->", round(np.std(extremo, ddof=1), 1))
```

## 4. Tres testigos de la dispersión

```python
from scipy import stats
q1, q3 = v.quantile(.25), v.quantile(.75)
print("Desviación estándar:", round(v.std(), 2))
print("Rango intercuartílico:", round(q3 - q1, 2), " (", round(q1, 2), "a", round(q3, 2), ")")
print("MAD:", round(stats.median_abs_deviation(v), 1))
for c in ["Seconds of Use", "Subscription Length"]:
    print(c, ": media", round(df[c].mean(), 2), " mediana", df[c].median(), " asimetría", round(stats.skew(df[c]), 3))
```

## 5. Testigos falsos: atributos ordinales (RAE 1.2)

```python
print(pd.crosstab(df["Age Group"], df["Age"]))
print("'Edad media':", round(df["Age"].mean(), 1), " -> no describe a nadie")
ca = df["Charge Amount"]
print("Charge Amount: mediana", ca.median(), " media", round(ca.mean(), 2))
print(ca.value_counts().sort_index())
print((df.groupby("Charge Amount")["Churn"].mean() * 100).round(1))
```

## 6. La historia aparece al comparar

```python
cols = ["Seconds of Use", "Customer Value", "Distinct Called Numbers", "Call Failure", "Subscription Length"]
print(df.groupby("Churn")[cols].median())
print((df.groupby("Complains")["Churn"].mean() * 100).round(1), df["Complains"].value_counts().to_dict())
df.boxplot(column="Seconds of Use", by="Churn"); plt.title("Quien se va usa mucho menos el servicio"); plt.suptitle(""); plt.show()
```

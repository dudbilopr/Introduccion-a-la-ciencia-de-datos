## 1. Importación de Librerías y Carga de Datos

Para este ejercicio, utilizaremos una base de datos real y pública sobre precios de viviendas en California (o similar) para observar cómo se distribuyen variables con alta asimetría.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuramos el estilo de las gráficas
sns.set_theme(style="whitegrid")

# Cargamos un dataset de ejemplo. Utilizaremos el popular dataset 'tips' (propinas en un restaurante)
# para entender ingresos y distribución.
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)

# Mostramos las primeras filas
df.head()
```

## 2. Medidas de Tendencia Central

La tendencia central busca el "centro" de los datos. 
- **Media (mean):** Promedio aritmético.
- **Mediana (median):** El valor central ordenado (50%).
- **Moda (mode):** El valor que más se repite.

```python
# Analizaremos la variable 'total_bill' (Total de la cuenta)

media = df['total_bill'].mean()
mediana = df['total_bill'].median()
moda = df['total_bill'].mode()[0] # mode() retorna una Serie, tomamos el primer valor

print(f"Media de la cuenta: ${media:.2f}")
print(f"Mediana de la cuenta: ${mediana:.2f}")
print(f"Moda de la cuenta: ${moda:.2f}")

print("\n--- Análisis ---")
print("Observa cómo la Media es mayor que la Mediana. Esto indica que hay unas pocas cuentas muy altas que están tirando del promedio hacia arriba.")
```

## 3. Medidas de Dispersión

¿Qué tan esparcidos están nuestros datos?
- **Varianza (var):** Diferencia al cuadrado respecto a la media.
- **Desviación Estándar (std):** Raíz cuadrada de la varianza. Nos da la medida en la misma unidad que la original.

```python
varianza = df['total_bill'].var()
desviacion_estandar = df['total_bill'].std()

print(f"Varianza: {varianza:.2f}")
print(f"Desviación Estándar: ${desviacion_estandar:.2f}")

# Podemos generar un resumen estadístico muy rápido usando .describe()
df.describe()
```

## 4. Medidas de Posición (Cuartiles y Percentiles)

Dividen nuestro conjunto ordenado en partes iguales. Útiles para detectar atípicos (outliers).

```python
# Percentiles con numpy
q1 = np.percentile(df['total_bill'], 25)
q2 = np.percentile(df['total_bill'], 50) # Equivalente a la mediana
q3 = np.percentile(df['total_bill'], 75)

print(f"Primer Cuartil (25% de los datos están por debajo de): ${q1:.2f}")
print(f"Segundo Cuartil (Mediana): ${q2:.2f}")
print(f"Tercer Cuartil (75% de los datos están por debajo de): ${q3:.2f}")

rango_intercuartilico = q3 - q1
print(f"\nRango Intercuartílico (IQR): ${rango_intercuartilico:.2f}")
```

## 5. Visualización de la Distribución
Una imagen vale más que mil métricas.

```python
plt.figure(figsize=(10, 5))

# Histograma
sns.histplot(df['total_bill'], kde=True, color='#003865')

# Líneas para media y mediana
plt.axvline(media, color='#E85C0B', linestyle='--', label=f'Media: {media:.1f}')
plt.axvline(mediana, color='green', linestyle='-', label=f'Mediana: {mediana:.1f}')

plt.title('Distribución del Total de la Cuenta')
plt.legend()
plt.show()
```

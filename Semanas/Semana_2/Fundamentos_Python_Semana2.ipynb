### 1️⃣ Variables y Tipos de Datos
En Python, no necesitamos declarar el tipo de variable. El intérprete lo asigna automáticamente (Tipado dinámico).

```python
# Variables Numéricas: Enteros (int)
variable_1 = 1
print("El valor es:", variable_1)
print("El tipo de dato es:", type(variable_1))
```

```python
# Variables Numéricas: Decimales (float)
variable_2 = 2.0
print("El valor es:", variable_2)
print("El tipo de dato es:", type(variable_2))
```

```python
# Variables de Texto: Cadenas (str)
variable_3 = 'a'
print("El valor es:", variable_3)
print("El tipo de dato es:", type(variable_3))
```

```python
# Variables Booleanas: Verdadero o Falso (bool)
variable_7 = True
print("El valor es:", variable_7)
print("El tipo de dato es:", type(variable_7))
```

---
### 2️⃣ Estructuras de Datos Básicas
A menudo necesitamos almacenar múltiples valores. Para esto, Python ofrece Listas, Tuplas y Conjuntos.

```python
# Lista (list): Mutables y ordenadas
variable_4 = [1, 2]
print("Lista:", variable_4, "| Tipo:", type(variable_4))

# Tupla (tuple): Inmutables y ordenadas
variable_5 = (1, '2')
print("Tupla:", variable_5, "| Tipo:", type(variable_5))

# Conjunto (set): Mutables pero desordenados y sin duplicados
variable_6 = {1.0, 2, 'a'}
print("Conjunto:", variable_6, "| Tipo:", type(variable_6))
```

### 3️⃣ Operaciones Matemáticas
Podemos operar los números usando los símbolos tradicionales.

```python
variable_a = 7.0
variable_b = float('3.0') # Convertimos un texto a decimal

suma = variable_a + variable_b
multiplicacion = variable_a * variable_b
division = variable_a / variable_b
modulo = variable_a % variable_b      # El residuo de la división
exponente = variable_a ** variable_b   # Potenciación
division_entera = variable_a // variable_b

print(f"Suma: {suma}")
print(f"Multiplicación: {multiplicacion}")
print(f"División: {division}")
print(f"Módulo (residuo): {modulo}")
print(f"Exponente: {exponente}")
print(f"División Entera: {division_entera}")
```

### 4️⃣ Funciones y Control de Flujo
Para no repetir código, empaquetamos nuestra lógica en **funciones** (usando `def`). Además, usamos `if`, `elif` y `else` para tomar decisiones.

```python
# Definimos funciones simples
def suma(a, b):
  return a + b

def resta(a, b):
  return a - b

def multiplicacion(a, b):
  return a * b

# Probamos las funciones
print("Suma (1+2):", suma(1, 2))
print("Resta (1-2):", resta(1, 2))
print("Multiplicación (1*2):", multiplicacion(1, 2))
```

```python
# Función maestra con control de flujo (IF / ELIF)
def calculadora(a, b, operacion):
  if operacion == 'suma':
    return suma(a, b)
  elif operacion == 'resta':
    return resta(a, b)
  elif operacion == 'multiplicacion':
    return multiplicacion(a, b)
  else:
    return "Operación no reconocida."

# Probamos la calculadora
res_suma = calculadora(1, 2, 'suma')
res_resta = calculadora(1, 2, 'resta')
res_mult = calculadora(1, 2, 'multiplicacion')

print(f"Resultados: {res_suma}, {res_resta}, {res_mult}")
```

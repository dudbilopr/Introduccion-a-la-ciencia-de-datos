## Estación 2. NumPy: sumar sin ciclos

```python
df = pd.read_csv(BASE + "bicicletas_dia.csv", parse_dates=["dteday"])
lista = df["cnt"].tolist()
total = 0
for v in lista:          # un ciclo: 731 vueltas
    total += v
arr = df["cnt"].to_numpy()
print("Con ciclo:", total, " con NumPy:", arr.sum())
print("Tipo:", arr.dtype, " bytes:", arr.nbytes)
print("Primeros 12 mayores que 5000:", arr[:12] > 5000)
```

## Estación 3. pandas: las cinco preguntas

```python
print(df.shape)
print(df.dtypes.value_counts())
print(df.head(3)[["dteday", "season", "temp", "cnt"]])
print("Vacíos:", df.isna().sum().sum())
print(df["cnt"].describe().round(2))
```

## Estación 3b. Filtrar y agrupar

```python
mask = df["cnt"] > 5000
print("Días con más de 5000:", mask.sum(), "=", round(mask.mean() * 100, 1), "%")
print(df.groupby("season")["cnt"].mean().round().astype(int))
dia_max = df.loc[df["cnt"].idxmax()]
print("Máximo:", dia_max["dteday"].date(), dia_max["cnt"])
```

## Estación 4. Distancia entre dos días (RAE 1.1)
El día de más alquileres contra el de menos, en temperatura (°C), humedad (%) y viento (km/h).

```python
df["temp_c"] = df["temp"] * 41
df["hum_pct"] = df["hum"] * 100
df["viento_kmh"] = df["windspeed"] * 67
cols = ["temp_c", "hum_pct", "viento_kmh"]
a = df.loc[df["cnt"].idxmax(), cols].to_numpy(float)
b = df.loc[df["cnt"].idxmin(), cols].to_numpy(float)
d = np.abs(a - b)
print("Diferencias:", d.round(2))
print("Manhattan:", round(d.sum(), 2), " Euclidiana:", round(np.sqrt((d ** 2).sum()), 2), " Suprema:", round(d.max(), 2))
print("Aporte de cada variable a la euclidiana al cuadrado (%):", (d ** 2 / (d ** 2).sum() * 100).round(1))
```

**La trampa de las unidades.** La humedad manda solo porque sus números son más grandes. Estandarice: reste la media y divida por la desviación estándar.

```python
z = (df[cols] - df[cols].mean()) / df[cols].std()
za, zb = z.loc[df["cnt"].idxmax()].to_numpy(), z.loc[df["cnt"].idxmin()].to_numpy()
print("Euclidiana estandarizada:", round(np.sqrt(((za - zb) ** 2).sum()), 3))
```

### 👩‍💻👨‍💻 Pair Programming: El Reto de Hoy
Vamos a simular ser los **Analistas de Datos** de una tienda local que acaba de abrir. Lamentablemente, no tienen un sistema que guarde el historial de ventas, así que vamos a **generar datos sintéticos** (inventados pero lógicos) usando Python puro, y luego utilizaremos **Pandas** para guardarlos, leerlos y analizarlos.

## Parte 1: El Piloto A toma el control ✈️
En esta sección generaremos los datos sintéticos usando Python nativo (`random` y listas).

```python
import random
from datetime import datetime, timedelta

# 1. Definimos los productos de nuestra tienda y sus precios base
productos = {
    'Laptops': 1500,
    'Smartphones': 800,
    'Audifonos': 150,
    'Teclados Mecanicos': 100,
    'Monitores': 300
}

# 2. Creamos listas vacías para guardar nuestros datos sintéticos
fechas_venta = []
productos_vendidos = []
cantidades = []
precios = []

# 3. Generamos 100 registros de ventas aleatorios
fecha_inicio = datetime.now() - timedelta(days=30) # Ventas del último mes

for i in range(100):
    # Seleccionamos un producto al azar
    producto = random.choice(list(productos.keys()))
    
    # Generamos una fecha aleatoria en los últimos 30 días
    dias_aleatorios = random.randint(0, 30)
    fecha = fecha_inicio + timedelta(days=dias_aleatorios)
    
    # Cantidad vendida (entre 1 y 5)
    cantidad = random.randint(1, 5)
    
    # Precio (el precio base más una pequeña variación)
    precio_final = productos[producto] + random.randint(-20, 20)
    
    # Agregamos los datos a las listas
    fechas_venta.append(fecha.strftime('%Y-%m-%d'))
    productos_vendidos.append(producto)
    cantidades.append(cantidad)
    precios.append(precio_final)

print("✅ Datos generados exitosamente. ¡Excelente trabajo Piloto A!")
```

## Parte 2: Cambio de Roles 🔄 (El Piloto B toma el control)
Ahora usaremos la librería **Pandas** para convertir estas listas sueltas en una tabla estructurada (DataFrame), guardarla en un archivo y analizarla.

```python
import pandas as pd

# 1. Creamos un diccionario con las listas que hizo el Piloto A
datos_diccionario = {
    'Fecha': fechas_venta,
    'Producto': productos_vendidos,
    'Cantidad': cantidades,
    'Precio_Unitario': precios
}

# 2. Convertimos el diccionario a un DataFrame de Pandas (Nuestra primera tabla estructurada)
df_ventas = pd.DataFrame(datos_diccionario)

# 3. Vamos a ver cómo luce nuestra tabla (los primeros 5 registros)
df_ventas.head()
```

### Guardando los datos (Persistencia)
En Ciencia de Datos, normalmente leemos datos de un archivo. Así que vamos a guardar nuestra tabla sintética como un archivo CSV (Valores separados por comas) aquí en Colab.

```python
# Guardamos el DataFrame a un archivo CSV. 
# 'index=False' evita que se guarde la columna numérica del lado izquierdo.
df_ventas.to_csv('ventas_sinteticas_tienda.csv', index=False)

print("✅ Archivo guardado como 'ventas_sinteticas_tienda.csv'. Puedes verlo en la carpeta de la izquierda en Colab.")
```

### Análisis Exploratorio Básico
Ahora vamos a leer ese mismo archivo que acabamos de guardar y haremos algunas preguntas básicas de analítica de negocios.

```python
# 1. Leemos el archivo CSV
df_analisis = pd.read_csv('ventas_sinteticas_tienda.csv')

# 2. Calculamos los ingresos totales por cada fila (Cantidad * Precio_Unitario)
df_analisis['Ingreso_Total'] = df_analisis['Cantidad'] * df_analisis['Precio_Unitario']

# 3. Respondemos: ¿Cuál fue el producto más vendido en términos de cantidad?
producto_mas_vendido = df_analisis.groupby('Producto')['Cantidad'].sum().sort_values(ascending=False)
print("--- Productos más vendidos (Unidades) ---")
print(producto_mas_vendido)
print("\n")

# 4. Respondemos: ¿Cuánto dinero generó la tienda en este mes simulado?
ingreso_total = df_analisis['Ingreso_Total'].sum()
print(f"--- Ingreso Total de la Tienda ---")
print(f"${ingreso_total:,.2f}")
```

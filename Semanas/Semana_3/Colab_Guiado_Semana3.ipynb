### 👨‍💻 Pair Programming: El Analista de Supermercado
En esta sesión, la gerencia del supermercado nos ha entregado un registro de ventas caótico. Debemos usar **Ciclos (for/while)** para limpiar ciertos errores y **Pandas** para extraer Insights valiosos.

## Parte 1: Generación y Limpieza con Ciclos (Piloto A) 🚀

```python
import random

# 1. Generamos datos 'sucios' para simular un sistema defectuoso
productos_sistema = ['  Manzanas', 'LECHE  ', 'pan', '   Queso', 'hUeVoS', ' CARNE']
ventas_sucias = [random.choice(productos_sistema) for _ in range(100)]

print("Muestra de datos sucios:")
print(ventas_sucias[:10])
```

```python
# 2. Limpiamos los datos usando un Ciclo FOR
ventas_limpias = []

for articulo in ventas_sucias:
    # strip() quita espacios en blanco, capitalize() pone la primera en mayúscula
    articulo_limpio = articulo.strip().capitalize()
    ventas_limpias.append(articulo_limpio)

print("\nMuestra de datos limpios:")
print(ventas_limpias[:10])
```

## Parte 2: El Poder de Pandas (Piloto B) 🐼
Cambio de roles. Ahora el Piloto B creará un DataFrame robusto con precios, sucursales y fechas.

```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# 1. Construimos nuestro diccionario maestro
sucursales = ['Centro', 'Norte', 'Sur']
precios_base = {'Manzanas': 1.5, 'Leche': 2.0, 'Pan': 1.0, 'Queso': 4.5, 'Huevos': 3.0, 'Carne': 8.0}

datos_supermercado = {
    'Fecha': [datetime.now() - timedelta(days=random.randint(0, 30)) for _ in range(100)],
    'Sucursal': [random.choice(sucursales) for _ in range(100)],
    'Producto': ventas_limpias,
    'Cantidad': [random.randint(1, 10) for _ in range(100)]
}

# 2. Convertimos a DataFrame
df = pd.DataFrame(datos_supermercado)

# 3. Añadimos el precio usando la función 'map' de Pandas conectada a nuestro diccionario
df['Precio_Unitario'] = df['Producto'].map(precios_base)

# 4. Calculamos el Total
df['Total_Venta'] = df['Cantidad'] * df['Precio_Unitario']

df.head()
```

## Parte 3: Analítica y Agrupaciones Avanzadas 📊
Juntos, respondamos las preguntas de negocio.

```python
# Pregunta 1: ¿Cuál es la sucursal que genera más dinero?
ventas_por_sucursal = df.groupby('Sucursal')['Total_Venta'].sum().sort_values(ascending=False)

print("--- Ventas Totales por Sucursal ---")
print(ventas_por_sucursal)
print("\n")

# Pregunta 2: ¿Cuántas ventas (transacciones) superaron los $20 dólares?
# Aquí usamos Filtrado Condicional en Pandas
ventas_altas = df[df['Total_Venta'] > 20]
cantidad_ventas_altas = len(ventas_altas)

print(f"Hubo {cantidad_ventas_altas} transacciones que superaron los $20.")
ventas_altas.head()
```
